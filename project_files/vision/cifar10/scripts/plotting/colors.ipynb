```python
import matplotlib.pyplot as plt
import scienceplots

# Set the style to 'science'
plt.style.use(["science", "vibrant"])

color_cycle = plt.rcParams['axes.prop_cycle'].by_key()['color']
```

```python
# visualize the color cycle
fig, ax = plt.subplots()
for i, color in enumerate(color_cycle):
    ax.plot([0, 1], [i, i], color=color, lw=5)
ax.set_yticks(range(len(color_cycle)))
ax.set_yticklabels([f'Color {i}' for i in range(len(color_cycle))])
ax.set_title('Color Cycle in Science Plot Style')
ax.set_xlabel('X-axis')
ax.set_ylabel('Color Index')
plt.grid(False)
plt.show()
# Save the figure
fig.savefig('color_cycle_science_style.pdf', dpi=300, bbox_inches='tight')
```
